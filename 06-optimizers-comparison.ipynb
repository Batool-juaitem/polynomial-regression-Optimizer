 Conclusion

In this project, five optimization algorithms (GD, SGD, Mini-Batch GD, RMSProp, and Adam) were implemented and compared using Polynomial Linear Regression on the California Housing Prices dataset.

All optimizers were trained using Momentum to improve the optimization process and accelerate convergence.

Based on the experimental results, **Adam with Momentum achieved the best overall performance**, obtaining an R² score of 0.6667, an RMSE of 66,091.06, and an MAE of 46,812.38.

Mini-Batch Gradient Descent achieved the second-best performance, while standard Gradient Descent produced the lowest R² score.

The results demonstrate that adaptive optimization algorithms, especially Adam, can improve model performance compared to standard Gradient Descent.

Overall, this project highlights the importance of choosing an appropriate optimization algorithm for training machine learning models.
